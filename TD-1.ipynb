## 2. Variables

### 2.1 Création de variable
Comme vu en cours, une variable sert à stocker une information en mémoire lors de l'exécution du programme. Une variable est identifiée par son nom. 
En Python, le nom qu'on donne à une variable doit suivre quelques règles :
- Commencer par une lettre ou un underscore (_) : un nom de variable ne peut pas commencer par un chiffre.
- Contenir uniquement des lettres (même accentuées), des chiffres et des underscores (_) : les espaces et les caractères spéciaux comme -, @, ! ou . sont interdits.
- Ne pas être un mot-clé réservé de Python : par exemple, if, for, while, class, True ou None.
- Respecter la casse : Python distingue les majuscules des minuscules. Ainsi, age, Age et AGE désignent trois variables différentes.

Une variable est créée lorsque vous **affectez une valeur** en utilisant le signe égal (`=`), comme dans l'exemple ci-dessous.

> ⚠️ **Note :** Juste parce que le nom de votre variable obeit aux règles de Python, ça ne vaut pas dire qui est approprié pour autant. Essayer de donner à vos variables des données qui sont cohérents avec leur usage, pour que votre code soit compréhensible (par des humains)

```python
# Assigning values
x = 5
y = 3
name = "Alice"
```

### 2.2 Afficher le contenu d'une variable
 Pour afficher la valeur d'une variable, on peut utiliser la fonction **print**.

```python
print("x =", x)
print("y =", y)
print("name =", name)
```

Le type d'une variable peut changer si vous lui assignez une nouvelle valeur d'un autre type.

```python
x = "Smart Mobility"
type(x)
```

```python
# Examples
n = 10          # int
pi = 3.14       # float
is_smart = True # bool

print(type(n), type(pi), type(is_smart))
```

### 3.4 Nombres

Python reconnaît si les nombres sont des entiers (`int`) ou des nombres à virgule flottante (décimaux) (`float`).
> **Note :**
> * Si vous incluez un point décimal, le nombre est un `float` ; sinon, c'est un `int`.
> * Les virgules (`,`) ne peuvent pas être utilisées dans les nombres en code. Utilisez des underscores (`_`) pour rendre les grands nombres plus lisibles.

```python
n = 10          # int
a = 10.0        # float
pi = 3.14       # float

m1 = 1,000,000   # not one million, but a tuple!
m2 = 1_000_000   # one million

print(type(n), type (a), type(pi), type(m1), type(m2))
```

#### Opérations sur les nombres
Les opérations arithmétiques de base sont intuitive dans python, il suffit simplement d'écrire les expressions employant des nombres directement ou des variables

```python
n * 2
```

Il existe également des opérations spécifiques aux nombre int
- Division entière **//**
- Reste de la division entière (modulo) **%**

Enfin, l'opérateur **\*\*** permet de calculer des puissances

```python
print(5 // 2)
print(5 % 2)
print(5**2)
```

#### Comparaisons
Les nombres peuvent être comparés avec ces opérateurs :

Ces opérateurs retournent des valeurs booléennes (True ou False)

### 3.5 Chaines de caractères
Une variable en Python peut contenir une chaine de caractères (du texte). Pour ce faire, ce texte doit être délimité par des double-quotes ou quotes simples

```python
example_text = "a"
example_text = "b"
```

```python
million_euros = "{:,.2f} €".format(1_000_000)
million_euros
```

> **Note :** Si une chaîne contient des quotes, ils peuvent être échappés avec un antislash (`\`) ou en utilisant l'autre type de guillemets.


```python
'Paris'                   # single quotes
"Paris"                   # double quotes
"Île-de-France's region"  # double quotes to include single quote
'Île-de-France\'s region' # escaping single quote
```

#### Sensibilité à la casse
[⏫](#Notebook-Overview)

Python est **sensible à la casse** pour le traitement des chaînes. Pour une comparaison *insensible à la casse*, convertissez-les en minuscules d'abord.


```python
'Paris' == 'paris'
```

```python
city = 'Paris'
city.lower() == 'paris'
```

#### Caractères d'échappement
L'antislash (`\`) échappe les caractères spéciaux :
* `\n` → newline
* `\t` → tab
* `\\` → literal backslash

Exécutez l'exemple ci-dessous

```python
print('C:\Python\transport_data')
```

Le **\t** de transport n'est pas affiché, une tabulation est inséré à la place car c'est ainsi qu'il est intérprété. Le **\P** est bien affiché mais il y a un message d'erreur indiquant que ce n'est pas reconnu en tant que séquence d'échappement. Pour indiquer à Python qu'un **\** n'est pas là pour spécifier une commande spéciale mais de simplement l'afficher, il suffit de le précéder d'un autre **\**

```python
print('C:\\Python\\transport_data') 
```

> **Astuce:**
> * Utilisez les chaînes brutes (`r'...'`) pour ignorer les séquences d'échappement.
> * Sous Windows, vous pouvez aussi utiliser des slashs (`/`) dans les chemins pour éviter les problèmes d'échappement.

```python
print(r'C:\Python\transport_data')   # raw string, no escaping needed
```

#### Chaînes multi-lignes
Triple quotes (`"""..."""` or `'''...'''`) can span multiple lines.
> **Astuce :** Utilisez un antislash (`\`) à la fin d'une ligne pour éviter les sauts de ligne automatiques.

```python
print("""\
Usage: paris-data [OPTIONS]
    -h          Display this usage message
    -t °C       Set temperature in Celsius
    -r          Include regions
""")
```

#### Concaténation
Deux ou plusieurs littéraux de chaîne côte à côte sont automatiquement joints.\
Vous pouvez aussi utiliser l'opérateur `+` pour concaténer des chaînes.

```python
print('Île' ' de-France') # Île de-France

s1 = 'Seine '
s2 = 'et Marne'
s3 = s1 + s2
print(s3) # Seine et Marne
```

## 3. Commentaires
[⏫](#Notebook-Overview)

En Python, les commentaires commencent par le symbole dièse `#`. Tout ce qui suit `#` sur la même ligne est ignoré par l'interpréteur.
> **Note :**
> * `#` **à l'intérieur d'une chaîne** est traité comme du texte, pas comme un commentaire.
> * Une chaîne de caractères autonome (non assignée à une variable) peut agir comme un commentaire multi-lignes.

```python
# Full-line comment:
x = 1  # Inline comment

# The # symbol inside a string is not a comment:
text = "# This is part of a string, not a comment"
```

## 4. Instruction `if`
[⏫](#Notebook-Overview)

Les instructions conditionnelles permettent au programme de choisir entre différents chemins.\
Une instruction `if` peut être utilisée seule, ou avec `elif` (*else if*) et/ou `else`.

#### Blocs et indentation
Les **blocs de code** en Python sont définis par l'**indentation**.
* Les lignes consécutives avec la même indentation forment un bloc, exécuté séquentiellement.

* L'indentation peut se faire avec des **espaces** ou des **tabulations**, mais ne **les mélangez pas dans le même fichier**.
* Les lignes qui attendent un bloc (p. ex. après if, for, def) doivent se terminer par un **deux-points** (`:`).

```python
city = "Versailles"

cities = ["Paris", "Lyon", "Marseille"]
departments = ["75", "69", "13"]

if city in cities:
    print("Found", city, "in the list of cities")
elif city in departments:
    print("Found", city, "as a department")
else:
    print("Did not find", city)
```
