## Syntaxe Python

```python
# Commentaire monoligne, Tous ce qui est après un # est considéré comme un commentaire et n'est pas traité par l'intérpéteur Python 

'''
Le même principe s'appliques aux commentaires multilignes de cette forme
'''


birth = 1970 # on peut affecter un nombre à une variable simplement avec le symbole =
now = 2026 
name = "John" # Pour ce qui est du texte, il faut le mettre entre double-quotes (") ou single-quotes(')

age = now - birth

"""
Les commentaires multilignes peuvent aussi s'écrire de cette forme
La ligne ci-dessous sert a afficher un message à l'écran de l'utilisateur via la fonction print.
Pour l'instant, remarquez simplement ce qui est passé à la fonction (les éléments entre parenthèses), ceux qui font référence à des variables et ceux qui sont simplement du texte.
Essayer de modifier le code et regarder comment le script ce comporte. 
"""
print("Hello, my name is ", name, ", I am ", age, " years old")

if age >= 65:
    print("I am retired")
else:
    print("I am not retired yet")
```
